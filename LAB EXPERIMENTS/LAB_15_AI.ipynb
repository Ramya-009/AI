```python
def predict(outlook, humidity):
    if outlook == "Sunny":
        if humidity == "High":
            return "No"
        else:
            return "Yes"
    elif outlook == "Rain":
        return "No"
    else:
        return "Yes"

print("Prediction for Sunny/High:", predict("Sunny", "High"))
print("Prediction for Overcast/Normal:", predict("Overcast", "Normal"))
```
