```python
# Bird Knowledge Base

birds = ["sparrow", "eagle", "crow", "parrot", "penguin", "ostrich", "kiwi"]

cannot_fly = ["penguin", "ostrich", "kiwi"]


def can_fly(B):
    return B in birds and B not in cannot_fly


def check(B):
    if can_fly(B):
        print(f"{B} can fly")
    elif B in birds:
        print(f"{B} cannot fly")
    else:
        print(f"{B} is not in the knowledge base")


# Queries

print(can_fly("sparrow"))
print(can_fly("penguin"))

check("ostrich")
check("dog")

# Query: can_fly(B)
for B in birds:
    if can_fly(B):
        print("B =", B)
```
