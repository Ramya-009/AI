```python
from collections import deque

# Possible actions: (Missionaries, Cannibals) to move
LOADS = [(1, 0), (2, 0), (0, 1), (0, 2), (1, 1)]

def valid(m, c):
    # m: missionaries on left bank, c: cannibals on left bank
    if not (0 <= m <= 3 and 0 <= c <= 3):
        return False
    if m > 0 and m < c:
        return False
    rm, rc = 3 - m, 3 - c
    if rm > 0 and rm < rc:
        return False
    return True

def solve():
    # State representation: (m_left, c_left, boat_position)
    # boat_position: 1 for left bank, 0 for right bank
    start = (3, 3, 1)
    goal = (0, 0, 0)

    queue = deque([start])
    parent = {start: None}

    while queue:
        cur = queue.popleft()
        if cur == goal:
            # Reconstruct path
            path = []
            while cur is not None:
                path.append(cur)
                cur = parent[cur]
            return path[::-1]

        m, c, b = cur
        for dm, dc in LOADS:
            if b == 1: # Boat is on left, moving to right
                nm, nc = m - dm, c - dc
            else: # Boat is on right, moving to left
                nm, nc = m + dm, c + dc

            nxt = (nm, nc, 1 - b)
            if valid(nm, nc) and nxt not in parent:
                parent[nxt] = cur
                queue.append(nxt)
    return None

path = solve()
if path:
    print("Steps:", len(path) - 1)
    for s in path:
        boat_side = "left" if s[2] == 1 else "right"
        print("Left bank: M=%d C=%d | Boat: %s | Right bank: M=%d C=%d" %
              (s[0], s[1], boat_side, 3 - s[0], 3 - s[1]))
else:
    print("No solution found.")
```
