```python
def minimax(depth, node, maximizing, values):
    if depth == 2:
        return values[node]

    if maximizing:
        return max(minimax(depth+1, node*2+i, False, values) for i in range(2))
    else:
        return min(minimax(depth+1, node*2+i, True, values) for i in range(2))

leaves = [3, 5, 2, 9]
print("Best value:", minimax(0, 0, True, leaves))
```
