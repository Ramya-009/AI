```python
regions = ['A', 'B', 'C']
neighbors = {'A': ['B', 'C'], 'B': ['A', 'C'], 'C': ['A', 'B']}
colors = ['Red', 'Green', 'Blue']
assigned = {}

def solve(i=0):
    if i == len(regions):
        return True
    region = regions[i]
    for color in colors:
        if all(assigned.get(n) != color for n in neighbors[region]):
            assigned[region] = color
            if solve(i + 1):
                return True
            del assigned[region]
    return False

solve()
print("Map colouring:", assigned)
```
