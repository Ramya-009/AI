```python
def vacuum_cleaner(room):
    position = 0
    for i in range(len(room)):
        if room[position] == 1:
            print("Room", position + 1, "is dirty. Cleaning...")
            room[position] = 0
        else:
            print("Room", position + 1, "is clean.")
        position += 1
    print("Final state:", room)

rooms = [1, 0, 1]
vacuum_cleaner(rooms)
```
